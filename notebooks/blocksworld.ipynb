# Blocksworld: sequential tasks without a planner

Ten blocks (numbered 0 to 9) stand on a table. A robot can **put** a clear block on another clear block, or **take** a block off another one.
Classic approaches search for a *plan*, a sequence of such actions. AICON does not plan. It follows the gradient of the goal cost through its components at every step, and the necessary intermediate steps (for example, first clearing a block) emerge from that.

**State.** AICON represents the world as a 10 × 10 matrix `below`: `below[i, j] = 1` means block `j` is somewhere below block `i`.

**Start (`3towers`).** Towers `0-1-2`, `4-3`, `6-7-8-9` (bottom to top) and block `5` alone.

**Goal (`StackAonBonC`).** Cost `2 − below[0, 4] − below[6, 0]`: block 0 must be on block 4 **and** block 6 on block 0. The robot therefore has to unstack three towers before it can build the new one, and nobody tells it that.

## Run the experiment

This is the same experiment as `aicon/blocksworld_experiment/run_experiment_sync.py`: up to 200 time steps, stopping when the goal is fulfilled.
AICON's detailed console output is collected in `result.log` instead of flooding the notebook. The run takes a short while on the CPU.

````python
INIT_SETUP = "3towers"      # "3towers" or "0towers"
GOAL = "StackAonBonC"       # "StackAonB", "StackAonBonC", "SmartStackAonBonC", "UnstackAonB"

result = run_blocksworld(init_setup=INIT_SETUP, goal=GOAL, max_steps=200)
result.summary()
````
