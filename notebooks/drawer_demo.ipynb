# Drawer opening in simulation

A simulated Franka Panda robot has to open a drawer (robosuite + MuJoCo). To do this it must locate the drawer handle, move there, grasp it, estimate how the drawer moves, and pull.
AICON does not execute these as separate, planned phases. All of them emerge from following the gradient of a single goal, *the drawer is open*, through a network of components:

| Type | Components |
|------|------------|
| Sensors | `EEPoseSensor`, `DrawerPoseSenser`, `EEForceSensor` |
| Estimators | `EEPoseEstimator`, `DrawerPositionEstimator`, `GraspedEstimator`, `KinematicJointEstimator` |
| Actions | `VeloEEAction` (end-effector velocity), `GripperAction` |
| Goal | `DrawerOpenViaJointGoal` |

The [step-by-step tutorial](https://github.com/tu-rbo/aicon/blob/master/Tutorial.md) builds exactly this system from scratch.
