# AICON: Active InterCONnect

Welcome to the AICON Virtual Research Lab. Everything here runs in your browser; nothing needs to be installed.

**AICON** is a framework for building robotic systems that estimate states and select actions by dynamically adjusting how their components interact, in response to feedback and regularities of the world.
Actions are chosen by **gradient descent** through a network of differentiable components. Subgoals that are encoded in these regularities are pursued implicitly, so sequential tasks can be solved **without an explicit plan**.

This lab accompanies the paper *No Plan but Everything Under Control: Robustly Solving Sequential Tasks with Dynamically Composed Gradient Descent* (Mengers & Brock, ICRA 2025). Videos and results: [project website](https://www.tu.berlin/robotics/papers/noplan).

## Key concepts

| Concept | What it does |
|---------|--------------|
| **Recursive estimators** (`EstimationComponent`) | Keep a differentiable estimate (and uncertainty) of a quantity of the world, updated recursively, e.g. with an EKF |
| **Sensors** (`SensorComponent`) | Provide measurements to the estimators |
| **Active interconnections** (`ActiveInterconnection`) | Differentiable relations between sensed, estimated and actuated quantities; which ones are active changes with the state |
| **Gradient-based actions** (`ActionComponent`) | Follow the steepest gradient of the goal cost through all components; subgoals emerge where the gradients lead |
